```
from pyomo.environ import *
from pyomo.opt import SolverFactory
```

## Définition du modèle

```
from pyomo.environ import *

model = ConcreteModel()

model.CLIENTS = Set(initialize=[1, 2, 3])
model.PRODUCTION = Set(initialize=[0, 1, 2, 3, 4, 5])
model.ARC = Set(dimen=2, initialize=[(i,j) for i in model.PRODUCTION for j in model.CLIENTS])
model.capacite = Param(model.PRODUCTION, model.CLIENTS, initialize={(0, 1): 0.0, (0, 2): 0.0, (0, 3): 0.0, (1, 1): 0.2, (1, 2): 0.4, (1, 3): 0.2, (2, 1): 0.4, (2, 2): 0.8, (2, 3): 0.4, (3, 1): 0.6, (3, 2): 1.2, (3, 3): 0.6, (4, 1): 0.8, (4, 2): 1.2, (4, 3): 0.8, (5, 1): 1.0, (5, 2): 1.2, (5, 3): 1.0}, within=NonNegativeReals)
model.revenu = Param(model.PRODUCTION, model.CLIENTS, initialize={(0, 1): 0.0, (0, 2): 0.0, (0, 3): 0.0, (1, 1): 1.0, (1, 2): 1.0, (1, 3): 1.0, (2, 1): 2.0, (2, 2): 5.0, (2, 3): 3.0, (3, 1): 4.0, (3, 2): 0.0, (3, 3): 5.0, (4, 1): 0.0, (4, 2): 0.0, (4, 3): 6.0, (5, 1): 0.0, (5, 2): 0.0, (5, 3): 7.0}, within=NonNegativeReals)
model.x = Var(model.PRODUCTION, model.CLIENTS, domain=NonNegativeReals)
# Constraint 0 not translated (contains LINGO constructs):
#    Original LINGO: @SUM(ARC: capacite * x) <= 1
#    TODO: translate this constraint from LINGO (@SUM/@FOR/...) to Pyomo.
def rule_for_0(model, c):
    return sum(model.x[p,c] for p in model.PRODUCTION) <= 1
model.c_for_0 = Constraint(model.CLIENTS, rule=rule_for_0)
# FAILED to translate FOR-loop 1: Format @FOR non reconnu : @FOR(ARC: @BIN(())
# Original: @FOR(ARC: @BIN(())
model.obj = Objective(expr=sum(model.revenu[p,c] * model.x[p,c] for (p,c) in model.ARC), sense=maximize)
```

## Résolution du modèle

```
solver = SolverFactory('gurobi')  # change selon ce que tu as installé
result = solver.solve(model, tee=True)
print('Solver status:', result.solver.status)
print('Termination condition:', result.solver.termination_condition)
```

## Valeurs optimales des variables

```
for v in model.component_objects(Var, active=True):
    print(f'Variable set: {v}')
    for index in v:
        print(f'   {index} = {v[index].value}')
```
